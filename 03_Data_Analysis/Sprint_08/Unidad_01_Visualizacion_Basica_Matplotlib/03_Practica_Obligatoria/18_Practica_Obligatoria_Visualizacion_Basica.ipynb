#### #1.2  

Recrea la siguiente figura, para ello construye en apartados separados cada gráfico y luego haz un apartado en el que se junten los cuatro.  

Las figuras son:
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center"). (Colores: azul, gris, aceituna, marrón, con trasparencia)  

2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Bradley Hand ITC", tamaño 18, color azul acero o "steelblue")
3. El scatter de área de agua de un ciudad con su poblacion.
4. El boxplot de la altura de las ciudades.

```python
import matplotlib.pyplot as plt

top = (
    df.nlargest(4, "population_total")[["city", "population_total"]]
    .sort_values("population_total", ascending=False)
)

plt.figure(figsize=(8, 5))
bars = plt.bar(
    top["city"],
    top["population_total"],
    color=["blue", "gray", "olive", "brown"],
    alpha=0.6
)

ymax = top["population_total"].max()
plt.ylim(0, ymax * 1.15)

for b, v in zip(bars, top["population_total"]):
    plt.text(
        b.get_x() + b.get_width() / 2,
        b.get_height(),
        f"{int(v):,}",
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["area_total_sq_mi"], bins=50, color="steelblue")
plt.title("Distribución del área total", fontname="Bradley Hand ITC", fontsize=18)
plt.xlabel("Área total (sq mi)", fontname="Bradley Hand ITC", fontsize=18)
plt.ylabel("Frecuencia", fontname="Bradley Hand ITC", fontsize=18)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df["area_water_sq_mi"], df["population_total"])
plt.xlabel("Área de agua (sq mi)")
plt.ylabel("Población")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.boxplot(df["elevation_ft"], vert=True)
plt.ylabel("Altura (ft)")
plt.show()
```

```python
import matplotlib.pyplot as plt

top = (
    df.nlargest(4, "population_total")[["city", "population_total"]]
    .sort_values("population_total", ascending=False)
)

fig, ax = plt.subplots(2, 2, figsize=(14, 9))

bars = ax[0, 0].bar(
    top["city"],
    top["population_total"],
    color=["blue", "gray", "olive", "brown"],
    alpha=0.6
)

ymax = top["population_total"].max()
ax[0, 0].set_ylim(0, ymax * 1.15)

for b, v in zip(bars, top["population_total"]):
    ax[0, 0].text(
        b.get_x() + b.get_width() / 2,
        b.get_height(),
        f"{int(v):,}",
        ha="center",
        va="bottom"
    )

ax[0, 1].hist(df["area_total_sq_mi"], bins=50, color="steelblue")
ax[0, 1].set_title("Distribución del área total", fontname="Bradley Hand ITC", fontsize=18)
ax[0, 1].set_xlabel("Área total (sq mi)", fontname="Bradley Hand ITC", fontsize=18)
ax[0, 1].set_ylabel("Frecuencia", fontname="Bradley Hand ITC", fontsize=18)

ax[1, 0].scatter(df["area_water_sq_mi"], df["population_total"])
ax[1, 0].set_xlabel("Área de agua (sq mi)")
ax[1, 0].set_ylabel("Población")

ax[1, 1].boxplot(df["elevation_ft"])
ax[1, 1].set_ylabel("Altura (ft)")

plt.tight_layout()
plt.show()
```
