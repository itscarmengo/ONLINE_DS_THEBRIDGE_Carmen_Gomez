```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import DBSCAN, AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score

sns.set(style="whitegrid", palette="muted", font_scale=1.2)
plt.rcParams["figure.figsize"] = (10, 6)

df = pd.read_csv("./data/empowering_drinks.csv", sep='|')
df.head()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

sns.pairplot(df[numeric_cols])
plt.show()

sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[numeric_cols])
```

```python
dbscan = DBSCAN(eps=0.8, min_samples=3)  # Ajusta eps/min_samples según resultados
db_labels = dbscan.fit_predict(X_scaled)
df['DBSCAN_cluster'] = db_labels

sns.scatterplot(x=df[numeric_cols[0]], y=df[numeric_cols[1]], hue='DBSCAN_cluster', data=df, palette='Set2')
plt.show()
```

```python
linked = linkage(X_scaled, method='ward')
plt.figure(figsize=(12, 6))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=True)
plt.show()

agglo = AgglomerativeClustering(n_clusters=4)  # Ajustable entre 3 y 5
agglo_labels = agglo.fit_predict(X_scaled)
df['Agglomerative_cluster'] = agglo_labels

sns.scatterplot(x=df[numeric_cols[0]], y=df[numeric_cols[1]], hue='Agglomerative_cluster', data=df, palette='Set1')
plt.show()
```

```python
try:
    print("Silhouette DBSCAN:", silhouette_score(X_scaled, db_labels))
except:
    print("DBSCAN genera ruido, Silhouette no computable para algunos clusters")

print("Silhouette Agglomerative:", silhouette_score(X_scaled, agglo_labels))
print("Calinski-Harabasz Agglomerative:", calinski_harabasz_score(X_scaled, agglo_labels))
print("Davies-Bouldin Agglomerative:", davies_bouldin_score(X_scaled, agglo_labels))
```

 - DBSCAN puede detectar grupos irregulares y ruido, pero requiere ajustar eps y min_samples.
 - Agglomerative genera un dendrograma útil para decidir el número de clusters (3-5 según enunciado).
 - Comparación especulativa: DBSCAN puede separar prototipos muy distintos, Agglomerative ofrece jerarquía.
 - Se pueden analizar qué segmentos son más parecidos usando proximidad en el dendrograma.
